```python
1. '''Smart Library Circulation System
Design a console-based Smart Library Circulation System for a college department. The
system should manage books, students, and borrowing activity in an offline environment.
The application must be menu-driven and continuously run until the user chooses to exit.
Users should be able to add new books with details such as book ID, title, author, total
copies, and available copies. Students must be registered using a unique student ID and
name.
The system should allow issuing books to students while enforcing rules such as:
• A book can be issued only if copies are available.
• A student can issue only a limited number of books.
• The same book cannot be issued twice to the same student.
Returning a book should update availability and calculate overdue penalties based on the
number of days exceeded. The program should also support searching books using partial
string matching and displaying overdue reports.
Appropriate data structures like dictionaries, lists, tuples, and strings must be used. Input
validation and logical consistency are mandatory'''
```

```python
# Simple Smart Library System (Beginner)

books = {}
students = {}
issued = {}

MAX_BOOKS = 2

while True:
    print("\n----- LIBRARY MENU -----")
    print("1. Add Book")
    print("2. Register Student")
    print("3. Issue Book")
    print("4. Return Book")
    print("5. Search Book")
    print("6. Exit")

    choice = input("Enter choice: ")

    # 1. Add Book
    if choice == "1":
        book_id = input("Book ID: ")
        title = input("Title: ")
        author = input("Author: ")
        copies = int(input("Total Copies: "))

        books[book_id] = [title, author, copies]
        print("Book added successfully")

    # 2. Register Student
    elif choice == "2":
        sid = input("Student ID: ")
        name = input("Student Name: ")

        students[sid] = name
        issued[sid] = []
        print("Student registered")

    # 3. Issue Book
    elif choice == "3":
        sid = input("Student ID: ")
        book_id = input("Book ID: ")

        if sid not in students:
            print("Student not found")

        elif book_id not in books:
            print("Book not found")

        elif books[book_id][2] == 0:
            print("No copies available")

        elif book_id in issued[sid]:
            print("Book already issued to this student")

        elif len(issued[sid]) >= MAX_BOOKS:
            print("Book limit reached")

        else:
            issued[sid].append(book_id)
            books[book_id][2] -= 1
            print("Book issued successfully")

    # 4. Return Book
    elif choice == "4":
        sid = input("Student ID: ")
        book_id = input("Book ID: ")

        if book_id in issued.get(sid, []):
            issued[sid].remove(book_id)
            books[book_id][2] += 1
            print("Book returned")

        else:
            print("Book not issued to student")

    # 5. Search Book
    elif choice == "5":
        key = input("Enter title keyword: ").lower()

        for b in books:
            if key in books[b][0].lower():
                print(b, books[b])

    # 6. Exit
    elif choice == "6":
        print("Exiting Library System")
        break

    else:
        print("Invalid choice")
```

```python
10. '''Log File Analyzer
Develop a Log File Analyzer for simulated system logs.
The program should parse text-based log entries, categorize events, count error types,
analyze user activity, and filter logs based on time or severity.
String splitting, pattern matching, and dictionary-based aggregation must be used. The
application should present analytical summaries via a menu-driven interface.'''
```

```python

# Simple Log File Analyzer (Beginner Level)

logs = []

while True:
    print("\n----- LOG FILE ANALYZER -----")
    print("1. Add Log Entry")
    print("2. Count Severity Types")
    print("3. Count Error Types")
    print("4. User Activity Report")
    print("5. Filter by Severity")
    print("6. Exit")

    choice = input("Enter choice: ")

    # 1. Add Log Entry
    if choice == "1":
        log = input("Enter log (DATE SEVERITY MESSAGE): ")
        logs.append(log)
        print("Log added")

    # 2. Count Severity Types
    elif choice == "2":
        severity_count = {}

        for log in logs:
            parts = log.split()
            severity = parts[1]

            if severity in severity_count:
                severity_count[severity] += 1
            else:
                severity_count[severity] = 1

        print("Severity Count:")
        for s in severity_count:
            print(s, ":", severity_count[s])

    # 3. Count Error Types
    elif choice == "3":
        error_count = {}

        for log in logs:
            if "ERROR" in log:
                parts = log.split()
                error = parts[2]

                if error in error_count:
                    error_count[error] += 1
                else:
                    error_count[error] = 1

        print("Error Types:")
        for e in error_count:
            print(e, ":", error_count[e])

    # 4. User Activity Report
    elif choice == "4":
        user_activity = {}

        for log in logs:
            parts = log.split()

            if "User" in parts[2]:
                user = parts[2]

                if user in user_activity:
                    user_activity[user] += 1
                else:
                    user_activity[user] = 1

        print("User Activity:")
        for u in user_activity:
            print(u, ":", user_activity[u], "activities")

    # 5. Filter by Severity
    elif choice == "5":
        level = input("Enter severity (INFO/ERROR/WARNING): ")

        print("Filtered Logs:")
        for log in logs:
            if level in log:
                print(log)

    # 6. Exit
    elif choice == "6":
        print("Exiting Log Analyzer")
        break

    else:
        print("Invalid choice")
```
